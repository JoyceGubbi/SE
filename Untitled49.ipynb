```python
import pandas as pd

data = {
    'Name': ['Rupa', 'Sita', 'Geeta'],
    'Age': ['25', '30', '35'],
    'City': ['Bagalkot', 'Solapur', 'Gulbarga']
}

df = pd.DataFrame(data)
df['Age'] = df['Age'].astype(int)
df_sorted = df.sort_values(by='Age', ascending=False)
print("DataFrame sorted by Age in descending order:")
print(df_sorted)
```
